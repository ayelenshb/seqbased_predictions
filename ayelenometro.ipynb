# Ayelenómetro

Fill in the **configuration** cell and run the **run** cell. The code is in the `ayelenometro/` folder; see `README.md`.

```python
# Configuration
# Each case needs: the alignment (FASTA, with the phenotype in the ID of each sequence),
# a name, the phenotype name and the position of the phenotype in the ID split by "_"
# (for example, -1 if it is the last field: ">Human_0.8095" -> 0.8095).
# The other options have default values (see help(Case)).

from ayelenometro import Case, run_case

cases = [
    Case(
        fasta="HIV_FPV.fasta",
        name="HIV_FPV",
        phenotype="FPV-IP (%)",
        phenotype_index=-1,
        # Alphabet (amino acid grouping per position): if it already exists for this case it is used;
        # if it does not exist, True creates it (it may take a while) and False stops the run.
        create_alphabet=True,
        # Optional: positions reported in the literature, in alignment numbering.
        literature=[10, 20, 24, 32, 33, 46, 47, 50, 54, 73, 74, 76, 82, 84, 88, 89, 90],
        # Other options (default values):
        # max_gaps=0.5, seed=42, lambda_reg=1.0,
        # variants=["SS", "MC-SS", "MC-E", "AC-MC-SS", "AC-MC-E"],
        # replicates=10, train_fraction=0.9, n_controls=100, n_permutations=0,
        # results_dir="results",
    ),
]
```

```python
# Run: the results of each case are written to <results_dir>/<name>/
for case in cases:
    summary = run_case(case)
    display(summary.round(3))
```
